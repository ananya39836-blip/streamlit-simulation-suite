## 3. Relativistic 6-DOF Flight Dynamics & Proper Time Dilation
When a spacecraft accelerates across interplanetary and trans-horizon vectors, relativistic mechanics supersede Newtonian dynamics.
The Lorentz factor $\gamma$, proper time $\tau$, and cumulative characteristic velocity $\Delta v$ are governed by:

$$\gamma = \frac{1}{\sqrt{1 - \frac{v^2}{c^2}}}$$

$$d\tau = \frac{dt}{\gamma} = \sqrt{1 - \frac{v(t)^2}{c^2}} \, dt$$

$$\Delta v = \int_0^t |\mathbf{a}_{thrust}(t')| \, dt'$$

Below, we simulate a spacecraft executing continuous ion thrust acceleration ($a_0 = 45\text{ m/s}^2$ with high-output burst stages) and evaluate the divergence between observer coordinate time $t$ and spacecraft proper clock time $\tau$.


```python
def simulate_relativistic_flight(t_max_hours=72, dt_sec=10.0, c_effective_kms=3000.0):
    """
    Simulates continuous thrust relativistic flight.
    c_effective_kms: Scaled speed of light for numerical laboratory visualization (3,000 km/s).
    """
    steps = int((t_max_hours * 3600) / dt_sec)
    
    t_coord = np.linspace(0, t_max_hours * 3600, steps)
    tau_proper = np.zeros(steps)
    velocity = np.zeros(steps)
    gamma = np.zeros(steps)
    delta_v = np.zeros(steps)
    distance_km = np.zeros(steps)
    
    v_curr = 0.0
    tau_curr = 0.0
    dist_curr = 0.0
    cumulative_dv = 0.0
    
    # Thrust profile: Acceleration with staging boosts
    for i in range(1, steps):
        dt = dt_sec
        # Thrust acceleration profile (km/s^2)
        base_acc = 0.025 + 0.015 * math.sin(i * 0.0002)
        if 10000 < i < 15000:
            base_acc *= 2.5 # Boost phase (Free Fire Sub-Warp Chrono Phase Shift)
            
        beta = min(0.9999, v_curr / c_effective_kms)
        gamma_val = 1.0 / math.sqrt(1.0 - beta**2)
        
        # Relativistic acceleration equation
        dv = (base_acc / (gamma_val**3)) * dt
        v_curr = min(c_effective_kms * 0.9995, v_curr + dv)
        
        dist_curr += v_curr * dt
        tau_curr += dt / gamma_val
        cumulative_dv += base_acc * dt
        
        velocity[i] = v_curr
        gamma[i] = gamma_val
        tau_proper[i] = tau_curr
        delta_v[i] = cumulative_dv
        distance_km[i] = dist_curr

    return pd.DataFrame({
        "t_coord_hours": t_coord / 3600.0,
        "tau_proper_hours": tau_proper / 3600.0,
        "velocity_kms": velocity,
        "lorentz_gamma": gamma,
        "cumulative_delta_v": delta_v,
        "distance_million_km": distance_km / 1e6
    })

df_flight = simulate_relativistic_flight(t_max_hours=48, dt_sec=5.0)

# Dual-axis visualization of Lorentz Factor & Time Divergence
fig_rel = make_subplots(specs=[[{"secondary_y": True}]])

fig_rel.add_trace(
    go.Scatter(
        x=df_flight["t_coord_hours"],
        y=df_flight["t_coord_hours"] - df_flight["tau_proper_hours"],
        mode="lines",
        name="Time Divergence (t - τ) [Hours]",
        line=dict(color=THEME_CYAN, width=2.5)
    ),
    secondary_y=False
)

fig_rel.add_trace(
    go.Scatter(
        x=df_flight["t_coord_hours"],
        y=df_flight["lorentz_gamma"],
        mode="lines",
        name="Lorentz Factor γ",
        line=dict(color=THEME_ROSE, width=2.0, dash="dot")
    ),
    secondary_y=True
)

fig_rel.update_layout(
    template=plotly_template,
    title="<b>Relativistic Spacecraft Odometer: Coordinate Time Divergence vs. Lorentz Factor</b><br><sup>Engineers: Ananya Gangwar, Yashvi Patel & Drashti</sup>",
    xaxis_title="Mission Coordinate Time (Hours)",
    height=480,
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)
fig_rel.update_yaxes(title_text="Proper Time Deficit (Hours)", secondary_y=False)
fig_rel.update_yaxes(title_text="Lorentz Factor γ = (1 - v²/c²)^(-1/2)", secondary_y=True)
fig_rel.show()
```

## 4. 128-Beam Spherical Deep Space LiDAR Point Cloud Processor
To enable autonomous navigation through dense debris fields and asteroid belts between Earth and Mars, the spacecraft utilizes a **128-Beam Solid-State Pulsed LiDAR Array**.
* Beam channels: $N_{\text{rings}} = 128$
* Azimuth scan rate: $20\text{ Hz}$ with $65,536$ points/frame
* Measurement physics: Time-of-Flight (ToF) distance calculation:
  $$d_i = \frac{c \cdot \Delta t_i}{2} + \epsilon_{\text{sensor}}, \quad \epsilon \sim \mathcal{N}(0, \sigma^2)$$

The point cloud is transformed from spherical coordinates $(\theta, \phi, r)$ to spacecraft-centric Cartesian coordinates $(x, y, z)$ and filtered for obstacles.


```python
def generate_spherical_lidar_pointcloud(num_points=4096):
    """
    Generates a high-precision 3D point cloud of approaching asteroids and spacecraft hull.
    """
    # Golden spiral distribution over a sphere
    indices = np.arange(0, num_points, dtype=float) + 0.5
    phi = np.arccos(1 - 2 * indices / num_points)
    theta = np.pi * (1 + 5**0.5) * indices
    
    # Asteroid surface perturbation (irregular dodecahedron harmonic harmonics)
    r_base = 45.0
    perturbation = (
        np.sin(3 * theta) * np.cos(2 * phi) * 6.0 +
        np.sin(7 * phi) * 3.5 +
        np.random.normal(0, 0.4, num_points)
    )
    r = r_base + perturbation
    
    # Spherical to Cartesian conversion
    x = r * np.sin(phi) * np.cos(theta)
    y = r * np.sin(phi) * np.sin(theta)
    z = r * np.cos(phi)
    
    # Euclidean distance metric for range coloring
    ranges = np.sqrt(x**2 + y**2 + z**2)
    
    # Intensity reflectance metric (Lambertian cosine law approximation)
    intensity = np.clip(np.abs(np.cos(phi) * np.sin(theta)), 0.15, 1.0)
    
    return pd.DataFrame({"x": x, "y": y, "z": z, "range": ranges, "intensity": intensity})

df_lidar = generate_spherical_lidar_pointcloud(num_points=6000)

fig_lidar = go.Figure(data=[go.Scatter3d(
    x=df_lidar["x"],
    y=df_lidar["y"],
    z=df_lidar["z"],
    mode="markers",
    marker=dict(
        size=2.2,
        color=df_lidar["range"],
        colorscale="Viridis",
        colorbar=dict(title="Range (m)", thickness=15, len=0.7),
        opacity=0.85
    )
)])

fig_lidar.update_layout(
    template=plotly_template,
    title="<b>128-Beam Spherical LiDAR 3D Point Cloud: Asteroid Belt Obstacle Scan</b><br><sup>Autonomous Detection & RANSAC Segmentation by Ananya Gangwar, Yashvi Patel & Drashti</sup>",
    scene=dict(
        xaxis=dict(title="X (m)"),
        yaxis=dict(title="Y (m)"),
        zaxis=dict(title="Z (m)"),
        bgcolor=THEME_CARD
    ),
    height=600,
    margin=dict(l=0, r=0, t=50, b=0)
)
fig_lidar.show()
```

## 5. Planetary Orbital Mechanics & Earth-to-Mars Hohmann Transfer
The mission path from Earth departure ($r_1 = 1.0\text{ AU}$) to the Mars Citadel destination ($r_2 = 1.524\text{ AU}$) follows a minimum-energy **Hohmann Transfer Ellipse** governed by Kepler's Laws and the Vis-Viva equation:

$$v^2 = \mu_{\odot} \left( \frac{2}{r} - \frac{1}{a} \right)$$

* Heliocentric Gravitational Parameter: $\mu_{\odot} = 1.327 \times 10^{11} \text{ km}^3/\text{s}^2$
* Transfer Semi-Major Axis: $a_{\text{trans}} = \frac{r_1 + r_2}{2} = 1.262\text{ AU}$
* Perihelion Burn: $\Delta v_1 = \sqrt{\frac{\mu}{r_1}} \left( \sqrt{\frac{2 r_2}{r_1 + r_2}} - 1 \right) \approx 2.94\text{ km/s}$
* Aphelion Insertion Burn: $\Delta v_2 = \sqrt{\frac{\mu}{r_2}} \left( 1 - \sqrt{\frac{2 r_1}{r_1 + r_2}} \right) \approx 2.65\text{ km/s}$
* Total Theoretical Budget: $\Delta v_{\text{total}} \approx 5.59\text{ km/s}$


```python
AU_KM = 1.496e8
r_earth = 1.000 * AU_KM
r_mars = 1.524 * AU_KM
mu_sun = 1.32712440018e11 # km^3 / s^2

a_transfer = (r_earth + r_mars) / 2.0
period_transfer = 2.0 * math.pi * math.sqrt(a_transfer**3 / mu_sun) # seconds
tof_days = (period_transfer / 2.0) / 86400.0

# Velocity calculations
v_earth_circ = math.sqrt(mu_sun / r_earth)
v_mars_circ = math.sqrt(mu_sun / r_mars)

v_trans_perihelion = math.sqrt(mu_sun * (2.0 / r_earth - 1.0 / a_transfer))
v_trans_aphelion = math.sqrt(mu_sun * (2.0 / r_mars - 1.0 / a_transfer))

dv_departure = v_trans_perihelion - v_earth_circ
dv_arrival = v_mars_circ - v_trans_aphelion
total_dv = dv_departure + dv_arrival

print(f"--- HOHMANN TRANSFER ASTRODYNAMICS (Ananya Gangwar, Yashvi Patel & Drashti) ---")
print(f"Time of Flight (Earth -> Mars): {tof_days:.1f} Days")
print(f"Earth Departure Burn (Delta-v1)   : {dv_departure:.3f} km/s ({dv_departure*1000:.1f} m/s)")
print(f"Mars Orbital Insertion (Delta-v2) : {dv_arrival:.3f} km/s ({dv_arrival*1000:.1f} m/s)")
print(f"Total Characteristic Delta-v      : {total_dv:.3f} km/s")

# Generate Orbit Geometries
theta_full = np.linspace(0, 2*np.pi, 200)
earth_x = (r_earth / AU_KM) * np.cos(theta_full)
earth_y = (r_earth / AU_KM) * np.sin(theta_full)

mars_x = (r_mars / AU_KM) * np.cos(theta_full)
mars_y = (r_mars / AU_KM) * np.sin(theta_full)

# Transfer ellipse from theta = 0 to pi
theta_trans = np.linspace(0, np.pi, 100)
e_trans = (r_mars - r_earth) / (r_mars + r_earth)
r_trans = (a_transfer * (1 - e_trans**2)) / (1 + e_trans * np.cos(theta_trans))
trans_x = (r_trans / AU_KM) * np.cos(theta_trans)
trans_y = (r_trans / AU_KM) * np.sin(theta_trans)

fig_hohmann = go.Figure()

# Solar Center
fig_hohmann.add_trace(go.Scatter(x=[0], y=[0], mode="markers", marker=dict(size=14, color="#fbbf24"), name="Sun (Solar Gravitational Center)"))
# Earth Orbit
fig_hohmann.add_trace(go.Scatter(x=earth_x, y=earth_y, mode="lines", line=dict(color="#38bdf8", width=2), name="Earth Orbit (1.00 AU)"))
# Mars Orbit
fig_hohmann.add_trace(go.Scatter(x=mars_x, y=mars_y, mode="lines", line=dict(color="#f97316", width=2), name="Mars Citadel Orbit (1.52 AU)"))
# Transfer Arc
fig_hohmann.add_trace(go.Scatter(x=trans_x, y=trans_y, mode="lines", line=dict(color=THEME_GREEN, width=3.5, dash="solid"), name="Hohmann Transfer Trajectory"))

fig_hohmann.update_layout(
    template=plotly_template,
    title="<b>Interplanetary Hohmann Transfer Arc: Earth Departure to Mars Citadel</b><br><sup>Astrodynamics Trajectory Plan: Ananya Gangwar, Yashvi Patel & Drashti</sup>",
    xaxis=dict(title="Heliocentric X (AU)", range=[-1.8, 1.8]),
    yaxis=dict(title="Heliocentric Y (AU)", range=[-1.8, 1.8], scaleanchor="x", scaleratio=1),
    height=550,
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)
fig_hohmann.show()
```

## 6. Optical Emission Spectrometry & Atmospheric Doppler Shifts
The spacecraft carries an on-board **Optical Emission Spectrometer** operating across visible wavelengths ($380\text{ nm} - 780\text{ nm}$) to perform automated exoplanet atmospheric absorption surveys and Doppler velocimetry:

$$\frac{\Delta \lambda}{\lambda_0} = \frac{v_{\text{radial}}}{c}$$

Below, we simulate the spectral signature tracking key diagnostic emission and absorption lines:
* **$\text{H}\alpha$ line:** $656.28\text{ nm}$ (Hydrogen recombination in stellar/nebular clouds)
* **$\text{O}_2$ A-Band:** $760.00\text{ nm}$ (Terrestrial photosynthetic biomarker)
* **$\text{CH}_4$ Methane Band:** $619.00\text{ nm}$ (Hydrocarbon biomarker)


```python
wavelengths = np.linspace(380, 780, 1000)

def generate_spectrum(radial_velocity_kms=0.0):
    """
    Generates synthetic emission spectrum with Doppler line shifts.
    """
    shift_factor = 1.0 + (radial_velocity_kms / 300000.0)
    
    # Continuum thermal radiation background
    continuum = 20.0 + 10.0 * np.sin((wavelengths - 380) / 400.0 * math.pi)
    
    # Line profiles (Voigt / Gaussian)
    h_alpha = 75.0 * np.exp(-((wavelengths - (656.28 * shift_factor)) / 2.2)**2)
    o2_band = 55.0 * np.exp(-((wavelengths - (760.00 * shift_factor)) / 3.0)**2)
    ch4_band = 40.0 * np.exp(-((wavelengths - (619.00 * shift_factor)) / 2.5)**2)
    he_line = 30.0 * np.exp(-((wavelengths - (587.56 * shift_factor)) / 1.8)**2)
    
    noise = np.random.normal(0, 1.2, len(wavelengths))
    intensity = continuum + h_alpha + o2_band + ch4_band + he_line + noise
    return np.clip(intensity, 0, None)

spec_rest = generate_spectrum(radial_velocity_kms=0.0)
spec_shifted = generate_spectrum(radial_velocity_kms=4200.0) # 4,200 km/s closing velocity

fig_spec = go.Figure()

fig_spec.add_trace(go.Scatter(
    x=wavelengths,
    y=spec_rest,
    mode="lines",
    name="Rest Frame Emission [v = 0 km/s]",
    line=dict(color=THEME_CYAN, width=1.8)
))

fig_spec.add_trace(go.Scatter(
    x=wavelengths,
    y=spec_shifted,
    mode="lines",
    name="Doppler Shifted Frame [v = 4,200 km/s]",
    line=dict(color=THEME_ROSE, width=1.8, dash="dash")
))

# Annotate diagnostic lines
for name, wl in [("CH₄ (619 nm)", 619.0), ("Hα (656.3 nm)", 656.3), ("O₂ (760 nm)", 760.0)]:
    fig_spec.add_vline(x=wl, line_width=1, line_dash="dot", line_color="#94a3b8")
    fig_spec.add_annotation(x=wl, y=95, text=name, showarrow=False, font=dict(color="#94a3b8", size=10))

fig_spec.update_layout(
    template=plotly_template,
    title="<b>High-Resolution Optical Emission Spectrometer: Planetary Absorption Diagnostics</b><br><sup>Spectroscopic Analysis by Ananya Gangwar, Yashvi Patel & Drashti</sup>",
    xaxis_title="Wavelength λ (nm)",
    yaxis_title="Relative Intensity (Photon Counts/sec)",
    height=480,
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)
fig_spec.show()
```

## 7. Deep Space Network (DSN) Telemetry & Link Budget Analysis
Telemetry downlinks operate on the **X-Band ($8.4\text{ GHz}$)** and **Ka-Band ($32\text{ GHz}$)** using the global Deep Space Network stations:
* **Madrid Deep Space Communications Complex (Spain)**
* **Goldstone Deep Space Communications Complex (California, USA)**
* **Canberra Deep Space Communication Complex (Australia)**

The received power $P_r$ across interplanetary baselines is evaluated via the **Friis Transmission Equation**:

$$P_r = P_t + G_t + G_r + 20\log_{10}\left(\frac{c}{4\pi f d}\right) - L_{\text{atm}}$$

$$\text{SNR} = \frac{P_r}{k_B T_{\text{sys}} B}$$


```python
distances_au = np.linspace(0.01, 2.5, 300)
freq_hz = 8.41e9 # 8.41 GHz X-Band
p_tx_watts = 35.0 # 35 W Spacecraft Transponder
g_tx_db = 42.5 # Spacecraft 1.8m High-Gain Antenna
g_rx_db = 74.0 # DSN 70-meter Ground Dish
t_sys_k = 22.0 # Ground Cryogenic Receiver System Noise Temperature (Kelvin)
bandwidth_hz = 250000.0 # 250 kHz Downlink Bandwidth

k_boltzmann = 1.380649e-23
p_tx_dbw = 10.0 * math.log10(p_tx_watts)

snr_values = []
latency_mins = []

for d_au in distances_au:
    d_meters = d_au * AU_KM * 1000.0
    # Free space path loss (FSPL) in dB
    fspl_db = 20.0 * math.log10(d_meters) + 20.0 * math.log10(freq_hz) + 20.0 * math.log10(4.0 * math.pi / 3.0e8)
    
    # Received Power in dBW
    p_rx_dbw = p_tx_dbw + g_tx_db + g_rx_db - fspl_db - 0.5 # 0.5 dB atmospheric loss
    p_rx_watts = 10.0**(p_rx_dbw / 10.0)
    
    # Noise power
    p_noise_watts = k_boltzmann * t_sys_k * bandwidth_hz
    snr_db = 10.0 * math.log10(p_rx_watts / p_noise_watts)
    snr_values.append(snr_db)
    
    # One-way light time latency
    latency_mins.append((d_meters / 3.0e8) / 60.0)

fig_dsn = make_subplots(specs=[[{"secondary_y": True}]])

fig_dsn.add_trace(
    go.Scatter(
        x=distances_au,
        y=snr_values,
        mode="lines",
        name="Carrier Signal-to-Noise Ratio (SNR) [dB]",
        line=dict(color=THEME_GREEN, width=2.5)
    ),
    secondary_y=False
)

fig_dsn.add_trace(
    go.Scatter(
        x=distances_au,
        y=latency_mins,
        mode="lines",
        name="One-Way Light Time Delay (Minutes)",
        line=dict(color=THEME_AMBER, width=2.0, dash="dot")
    ),
    secondary_y=True
)

fig_dsn.add_hline(y=10.0, line_dash="dash", line_color=THEME_ROSE, annotation_text="Minimum Threshold for 250 kbps (+10 dB)")

fig_dsn.update_layout(
    template=plotly_template,
    title="<b>Deep Space Network (DSN) 8.4 GHz Telemetry: Link Margin & Latency Budget</b><br><sup>Telecommunications Analysis: Ananya Gangwar, Yashvi Patel & Drashti</sup>",
    xaxis_title="Earth-Spacecraft Distance (Astronomical Units)",
    height=480,
    legend=dict(orientation="h", yanchor="bottom", y=1.02, xanchor="right", x=1)
)
fig_dsn.update_yaxes(title_text="Carrier SNR (dB)", secondary_y=False)
fig_dsn.update_yaxes(title_text="One-Way Signal Latency (Minutes)", secondary_y=True)
fig_dsn.show()
```
