## 2. Operator Overloading & 3D Spatial Vector Mathematics (`Vector3D`)
In deep-space navigation, 3D vectors represent positions, velocities, acceleration vectors, and LiDAR ray intercepts.
Below, we construct an encapsulated **`Vector3D`** class that overloads:
* **Binary `+` and `-` operators:** Vector addition and subtraction.
* **Scalar `*` operator:** Scaling velocity and thrust vectors.
* **Compound `+=` operator:** In-place positional integration $\mathbf{r} \leftarrow \mathbf{r} + \mathbf{v} \Delta t$.
* **Relational `==` operator:** Precision tolerance equality check.
* **Stream insertion `<<` operator:** Formatted mission-control console serialization.


````c++
#include <iostream>
#include <vector>
#include <string>
#include <cmath>
#include <iomanip>
#include <memory>
#include <algorithm>
#include <sstream>

// ============================================================================
// ENCAPSULATED MATHEMATICAL CLASS: Vector3D
// Demonstrates Encapsulation & Comprehensive Operator Overloading
// ============================================================================
class Vector3D {
private:
    double x, y, z;

public:
    // Constructors (Constructor Overloading)
    Vector3D() : x(0.0), y(0.0), z(0.0) {}
    Vector3D(double px, double py, double pz) : x(px), y(py), z(pz) {}

    // Encapsulated Getters
    double getX() const { return x; }
    double getY() const { return y; }
    double getZ() const { return z; }

    // Magnitude & Normalization
    double magnitude() const {
        return std::sqrt(x * x + y * y + z * z);
    }

    Vector3D normalize() const {
        double mag = magnitude();
        if (mag < 1e-15) return Vector3D(0.0, 0.0, 0.0);
        return Vector3D(x / mag, y / mag, z / mag);
    }

    // ------------------------------------------------------------------------
    // OPERATOR OVERLOADING (Object Overloading)
    // ------------------------------------------------------------------------

    // Overload + operator (Vector + Vector)
    Vector3D operator+(const Vector3D& other) const {
        return Vector3D(x + other.x, y + other.y, z + other.z);
    }

    // Overload - operator (Vector - Vector)
    Vector3D operator-(const Vector3D& other) const {
        return Vector3D(x - other.x, y - other.y, z - other.z);
    }

    // Overload * operator (Vector * Scalar)
    Vector3D operator*(double scalar) const {
        return Vector3D(x * scalar, y * scalar, z * scalar);
    }

    // Overload += operator (In-place compound accumulation)
    Vector3D& operator+=(const Vector3D& other) {
        x += other.x;
        y += other.y;
        z += other.z;
        return *this;
    }

    // Overload == operator (Tolerance-aware equality)
    bool operator==(const Vector3D& other) const {
        const double eps = 1e-9;
        return (std::abs(x - other.x) < eps) &&
               (std::abs(y - other.y) < eps) &&
               (std::abs(z - other.z) < eps);
    }

    // Overload << operator (Stream insertion for formatted output)
    friend std::ostream& operator<<(std::ostream& os, const Vector3D& vec) {
        os << "[" << std::fixed << std::setprecision(3) 
           << vec.x << ", " << vec.y << ", " << vec.z << "] (mag: " 
           << vec.magnitude() << ")";
        return os;
    }
};

// Test Vector3D Operator Overloading
Vector3D posInitial(0.0, 40.0, 80.0);
Vector3D vel(0.0, 5.0, -15.0);
double dt = 2.0;

Vector3D posNext = posInitial + (vel * dt); // Uses operator*, operator+
posInitial += (vel * dt);                   // Uses operator+=

std::cout << "[OPERATOR OVERLOADING VERIFIED]\n";
std::cout << "Initial Position + (Vel * dt) = " << posNext << "\n";
std::cout << "Equality Check (posNext == posInitial): " 
          << ((posNext == posInitial) ? "TRUE (PASS)" : "FALSE") << "\n";
std::cout << "Vector Engine Developed by: Ananya Gangwar, Yashvi Patel & Drashti\n";
````

## 4. Abstract Interfaces & Method Overloading
Below, we define:
* **`ISimulationModule`:** Pure virtual interface guaranteeing dynamic polymorphic execution.
* **`SpacecraftSubsystem`:** Encapsulated base class demonstrating **Method Overloading** for power configuration:
  * `setPowerConsumption(double watts)`
  * `setPowerConsumption(double voltage, double current)` $\to P = V \times I$


````c++
// ============================================================================
// ABSTRACT INTERFACE: ISimulationModule
// Pure virtual contract for polymorphic runtime dispatch
// ============================================================================
class ISimulationModule {
public:
    virtual ~ISimulationModule() = default;

    // Pure virtual functions (Method Overriding enforced in derived classes)
    virtual void initialize() = 0;
    virtual void step(double dtSec) = 0;
    virtual void displayDiagnostics() const = 0;
    virtual std::string getModuleIdentifier() const = 0;
};

// ============================================================================
// ENCAPSULATED BASE CLASS: SpacecraftSubsystem
// Demonstrates Encapsulation & Method Overloading
// ============================================================================
class SpacecraftSubsystem : public ISimulationModule {
protected:
    std::string subsystemName;
    double powerConsumptionWatts;
    bool isOnline;

public:
    SpacecraftSubsystem(std::string name, double powerWatts)
        : subsystemName(std::move(name)), powerConsumptionWatts(powerWatts), isOnline(false) {}

    // Encapsulated Accessors
    const std::string& getSubsystemName() const { return subsystemName; }
    double getPowerConsumptionWatts() const { return powerConsumptionWatts; }
    bool getIsOnline() const { return isOnline; }
    void setOnlineState(bool state) { isOnline = state; }

    // ------------------------------------------------------------------------
    // METHOD OVERLOADING (Compile-time Polymorphism)
    // ------------------------------------------------------------------------
    
    // Overload 1: Direct wattage specification
    void setPowerConsumption(double watts) {
        powerConsumptionWatts = std::max(0.0, watts);
    }

    // Overload 2: Voltage & Current calculation (P = V * I)
    void setPowerConsumption(double voltageVolts, double currentAmperes) {
        powerConsumptionWatts = std::max(0.0, voltageVolts * currentAmperes);
    }

    std::string getModuleIdentifier() const override {
        return subsystemName + " [Subsystem]";
    }
};

std::cout << "[BASE SUBSYSTEM ARCHITECTURE READY]\n";
std::cout << "Encapsulated accessors & overloaded power methods operational.\n";
````

## 5. Derived Sensor Subsystems: Overriding & Overloading
We implement three concrete instruments inheriting from `SpacecraftSubsystem`:
1. **`LidarSensor`:** Demonstrates `generateScan()` method overloading and `override` on lifecycle methods.
2. **`OpticalSpectrometer`:** Demonstrates `setRadialVelocity()` method overloading (km/s vs. m/s).
3. **`DSNTransceiver`:** Demonstrates `setTargetDistance()` method overloading (AU vs. Kilometers).


````c++
// ============================================================================
// DERIVED SENSOR 1: LidarSensor
// Demonstrates Inheritance, Method Overriding & Method Overloading
// ============================================================================
class LidarSensor : public SpacecraftSubsystem {
private:
    int beamCount;
    int pointCapacity;
    std::vector<Vector3D> pointCloud;
    double meanRangeMeters;

public:
    LidarSensor(int beams, int capacity, double powerWatts)
        : SpacecraftSubsystem("128-Beam Spherical Solid-State LiDAR", powerWatts),
          beamCount(beams), pointCapacity(capacity), meanRangeMeters(0.0) {}

    // METHOD OVERRIDING (Runtime Polymorphism)
    void initialize() override {
        setOnlineState(true);
        generateScan();
    }

    void step(double dtSec) override {
        if (!getIsOnline()) return;
        generateScan();
    }

    // METHOD OVERLOADING 1: Standard procedural scan
    void generateScan() {
        generateScan(45.0, 1.0);
    }

    // METHOD OVERLOADING 2: Custom range & terrain jitter parameterization
    void generateScan(double baseRange, double jitterFactor) {
        pointCloud.clear();
        pointCloud.reserve(pointCapacity);

        const double goldenRatio = (1.0 + std::sqrt(5.0)) / 2.0;
        double sumRange = 0.0;

        for (int i = 0; i < pointCapacity; ++i) {
            double theta = 2.0 * M_PI * i / goldenRatio;
            double phi = std::acos(1.0 - 2.0 * (i + 0.5) / pointCapacity);

            double harmonic = (6.5 * std::sin(3.0 * theta) * std::cos(2.0 * phi) + 2.0 * std::sin(7.0 * phi)) * jitterFactor;
            double r = baseRange + harmonic;

            double px = r * std::sin(phi) * std::cos(theta);
            double py = r * std::sin(phi) * std::sin(theta);
            double pz = r * std::cos(phi);

            Vector3D pt(px, py, pz);
            pointCloud.push_back(pt);
            sumRange += pt.magnitude();
        }
        meanRangeMeters = sumRange / pointCapacity;
    }

    double getMeanRange() const { return meanRangeMeters; }
    size_t getPointCount() const { return pointCloud.size(); }

    void displayDiagnostics() const override {
        std::cout << "[LIDAR TELEMETRY] Status: " << (isOnline ? "ONLINE" : "STANDBY")
                  << " | Beams: " << beamCount 
                  << " | Points: " << pointCloud.size() 
                  << " | Mean Range: " << std::fixed << std::setprecision(2) << meanRangeMeters << " m"
                  << " | Power: " << powerConsumptionWatts << " W\n";
    }
};

// ============================================================================
// DERIVED SENSOR 2: OpticalSpectrometer
// Demonstrates Inheritance, Method Overriding & Doppler Velocity Overloading
// ============================================================================
struct DiagnosticLine {
    std::string element;
    double restWavelengthNm;
    double relativeIntensity;
};

class OpticalSpectrometer : public SpacecraftSubsystem {
private:
    double currentRadialVelocityKmS;
    std::vector<DiagnosticLine> lines;

public:
    explicit OpticalSpectrometer(double powerWatts)
        : SpacecraftSubsystem("High-Res Optical Emission Spectrometer", powerWatts),
          currentRadialVelocityKmS(0.0) {
        lines = {
            {"He I", 587.56, 30.0},
            {"CH4 (Methane)", 619.00, 42.0},
            {"H-alpha", 656.28, 85.0},
            {"O2 A-Band", 760.00, 60.0}
        };
    }

    // METHOD OVERLOADING 1: Velocity in km/s
    void setRadialVelocity(double vKmS) {
        currentRadialVelocityKmS = vKmS;
    }

    // METHOD OVERLOADING 2: Velocity in m/s
    void setRadialVelocity(double vMps, bool isMetersPerSecond) {
        if (isMetersPerSecond) {
            currentRadialVelocityKmS = vMps / 1000.0;
        } else {
            currentRadialVelocityKmS = vMps;
        }
    }

    void initialize() override {
        setOnlineState(true);
    }

    void step(double dtSec) override {
        // Real-time spectroscopy integration
    }

    void displayDiagnostics() const override {
        const double cLight = 299792.458;
        std::cout << "[SPECTROMETER TELEMETRY] Status: " << (isOnline ? "ONLINE" : "STANDBY")
                  << " | Closing Vel: " << std::fixed << std::setprecision(1) << currentRadialVelocityKmS << " km/s\n";
        for (const auto& line : lines) {
            double shifted = line.restWavelengthNm * (1.0 + currentRadialVelocityKmS / cLight);
            std::cout << "   -> Line: " << std::left << std::setw(15) << line.element
                      << "Rest: " << std::fixed << std::setprecision(2) << line.restWavelengthNm << " nm"
                      << " | Shifted: " << shifted << " nm"
                      << " | Δλ: +" << (shifted - line.restWavelengthNm) << " nm\n";
        }
    }
};

// ============================================================================
// DERIVED SENSOR 3: DSNTransceiver
// Demonstrates Inheritance, Method Overriding & Distance Unit Overloading
// ============================================================================
class DSNTransceiver : public SpacecraftSubsystem {
private:
    double carrierFrequencyHz;
    double txPowerWatts;
    double groundDishGainDb;
    double targetDistanceAu;

public:
    DSNTransceiver(double freqHz, double txWatts, double groundGainDb, double powerWatts)
        : SpacecraftSubsystem("Deep Space Network (DSN) Transponder", powerWatts),
          carrierFrequencyHz(freqHz), txPowerWatts(txWatts), 
          groundDishGainDb(groundGainDb), targetDistanceAu(1.0) {}

    // METHOD OVERLOADING 1: Target distance in Astronomical Units
    void setTargetDistance(double dAu) {
        targetDistanceAu = std::max(0.001, dAu);
    }

    // METHOD OVERLOADING 2: Target distance in Kilometers
    void setTargetDistance(double dKm, bool isKilometers) {
        const double auKm = 149597870.7;
        if (isKilometers) {
            targetDistanceAu = std::max(0.001, dKm / auKm);
        } else {
            targetDistanceAu = std::max(0.001, dKm);
        }
    }

    void initialize() override {
        setOnlineState(true);
    }

    void step(double dtSec) override {
        // Continuous carrier phase locking
    }

    void displayDiagnostics() const override {
        const double auKm = 149597870.7;
        const double kBoltzmann = 1.380649e-23;
        double dMeters = targetDistanceAu * auKm * 1000.0;
        
        double fsplDb = 20.0 * std::log10(dMeters) + 20.0 * std::log10(carrierFrequencyHz) + 20.0 * std::log10(4.0 * M_PI / 3.0e8);
        double pTxDbw = 10.0 * std::log10(txPowerWatts);
        double pRxDbw = pTxDbw + 42.5 + groundDishGainDb - fsplDb - 0.5;
        double pRxWatts = std::pow(10.0, pRxDbw / 10.0);
        double pNoiseWatts = kBoltzmann * 22.0 * 250000.0;
        double snrDb = 10.0 * std::log10(pRxWatts / pNoiseWatts);
        double latencyMinutes = (dMeters / 3.0e8) / 60.0;

        std::cout << "[DSN TRANSCEIVER TELEMETRY] Status: " << (isOnline ? "LINK ACTIVE" : "DISCONNECTED")
                  << " | Freq: " << (carrierFrequencyHz / 1e9) << " GHz"
                  << " | Dist: " << std::fixed << std::setprecision(2) << targetDistanceAu << " AU"
                  << " | Carrier SNR: " << std::setprecision(1) << snrDb << " dB"
                  << " | Latency: " << latencyMinutes << " min\n";
    }
};

std::cout << "[SENSOR SUBSYSTEMS COMPILED]\n";
std::cout << "Method overriding & method overloading verified for LiDAR, Spectrometer, and DSN.\n";
std::cout << "Specialists: Ananya Gangwar, Yashvi Patel & Drashti\n";
````

## 6. Trajectory Propagators: Orbital & Relativistic Kinematics
We derive concrete mathematical engines from `TrajectoryPropagator`:
* **`HohmannTransferSolver`:** Keplerian solver featuring overloaded `solveManeuvers()` methods.
* **`RelativisticKinematicsSolver`:** Numerical RK4 integrator featuring overloaded `step()` methods accepting variable thrust scales.


````c++
// ============================================================================
// BASE CLASS: TrajectoryPropagator
// ============================================================================
class TrajectoryPropagator : public ISimulationModule {
protected:
    std::string trajectoryType;
    double heliocentricMu;
    Vector3D currentPosition;
    Vector3D currentVelocity;

public:
    TrajectoryPropagator(std::string type, double mu)
        : trajectoryType(std::move(type)), heliocentricMu(mu),
          currentPosition(0.0, 0.0, 0.0), currentVelocity(0.0, 0.0, 0.0) {}

    const std::string& getTrajectoryType() const { return trajectoryType; }
    double getMu() const { return heliocentricMu; }
    const Vector3D& getPosition() const { return currentPosition; }
    const Vector3D& getVelocity() const { return currentVelocity; }

    std::string getModuleIdentifier() const override {
        return trajectoryType + " [Propagator]";
    }
};

// ============================================================================
// DERIVED PROPAGATOR 1: HohmannTransferSolver
// Demonstrates Inheritance, Method Overriding & Method Overloading
// ============================================================================
class HohmannTransferSolver : public TrajectoryPropagator {
private:
    double rDepartureKm;
    double rArrivalKm;
    double aTransferKm;
    double timeOfFlightDays;
    double deltaV1KmS;
    double deltaV2KmS;
    double totalDeltaVKmS;

public:
    HohmannTransferSolver(double rDepAu, double rArrAu, double mu)
        : TrajectoryPropagator("Heliocentric Hohmann Conic Transfer", mu) {
        const double auKm = 149597870.7;
        rDepartureKm = rDepAu * auKm;
        rArrivalKm = rArrAu * auKm;
        solveManeuvers();
    }

    void initialize() override {
        solveManeuvers();
    }

    void step(double dtSec) override {
        // Orbital propagation step
    }

    // METHOD OVERLOADING 1: Solve using current internal radii
    void solveManeuvers() {
        solveManeuvers(rDepartureKm, rArrivalKm);
    }

    // METHOD OVERLOADING 2: Solve with arbitrary custom planetary radii
    void solveManeuvers(double rDep, double rArr) {
        rDepartureKm = rDep;
        rArrivalKm = rArr;
        aTransferKm = (rDepartureKm + rArrivalKm) / 2.0;

        double periodSec = 2.0 * M_PI * std::sqrt(std::pow(aTransferKm, 3) / heliocentricMu);
        timeOfFlightDays = (periodSec / 2.0) / 86400.0;

        double vDepCirc = std::sqrt(heliocentricMu / rDepartureKm);
        double vArrCirc = std::sqrt(heliocentricMu / rArrivalKm);

        double vPerihelion = std::sqrt(heliocentricMu * (2.0 / rDepartureKm - 1.0 / aTransferKm));
        double vAphelion = std::sqrt(heliocentricMu * (2.0 / rArrivalKm - 1.0 / aTransferKm));

        deltaV1KmS = vPerihelion - vDepCirc;
        deltaV2KmS = vArrCirc - vAphelion;
        totalDeltaVKmS = deltaV1KmS + deltaV2KmS;
    }

    void displayDiagnostics() const override {
        std::cout << "[HOHMANN PROPAGATOR DIAGNOSTICS]\n"
                  << "   Trajectory Arc : Earth (1.0 AU) -> Mars Citadel (1.524 AU)\n"
                  << "   Time of Flight : " << std::fixed << std::setprecision(1) << timeOfFlightDays << " Days (" << (timeOfFlightDays / 30.4375) << " Months)\n"
                  << "   Departure Burn : " << std::setprecision(3) << deltaV1KmS << " km/s (" << (deltaV1KmS * 1000.0) << " m/s)\n"
                  << "   Insertion Burn : " << deltaV2KmS << " km/s (" << (deltaV2KmS * 1000.0) << " m/s)\n"
                  << "   Total Δv Budget: " << totalDeltaVKmS << " km/s\n";
    }
};

// ============================================================================
// DERIVED PROPAGATOR 2: RelativisticKinematicsSolver
// Demonstrates Inheritance, Method Overriding & Step Method Overloading
// ============================================================================
class RelativisticKinematicsSolver : public TrajectoryPropagator {
private:
    double cEffectiveKmS;
    double coordinateTimeHours;
    double properTimeHours;
    double currentVelocityKmS;
    double lorentzGamma;
    double cumulativeDeltaVKmS;
    double distanceKm;

public:
    RelativisticKinematicsSolver(double cEffective, double mu)
        : TrajectoryPropagator("Relativistic 6-DOF Kinematics Engine", mu),
          cEffectiveKmS(cEffective), coordinateTimeHours(0.0), properTimeHours(0.0),
          currentVelocityKmS(0.0), lorentzGamma(1.0), cumulativeDeltaVKmS(0.0), distanceKm(0.0) {}

    void initialize() override {
        coordinateTimeHours = 0.0;
        properTimeHours = 0.0;
        currentVelocityKmS = 0.0;
        lorentzGamma = 1.0;
        cumulativeDeltaVKmS = 0.0;
        distanceKm = 0.0;
    }

    // METHOD OVERLOADING 1: Standard step using baseline 35 m/s^2 ion thrust
    void step(double dtSec) override {
        step(dtSec, 0.035);
    }

    // METHOD OVERLOADING 2: Custom step with configurable acceleration scaling
    void step(double dtSec, double customThrustKmS2) {
        double beta = std::min(0.9999, currentVelocityKmS / cEffectiveKmS);
        lorentzGamma = 1.0 / std::sqrt(1.0 - beta * beta);

        // Relativistic acceleration
        double dv = (customThrustKmS2 / (lorentzGamma * lorentzGamma * lorentzGamma)) * dtSec;
        currentVelocityKmS = std::min(cEffectiveKmS * 0.9995, currentVelocityKmS + dv);

        distanceKm += currentVelocityKmS * dtSec;
        coordinateTimeHours += dtSec / 3600.0;
        properTimeHours += (dtSec / lorentzGamma) / 3600.0;
        cumulativeDeltaVKmS += customThrustKmS2 * dtSec;
    }

    void displayDiagnostics() const override {
        double timeDeficit = coordinateTimeHours - properTimeHours;
        std::cout << "[RELATIVISTIC KINEMATICS DIAGNOSTICS]\n"
                  << "   Coordinate Time : " << std::fixed << std::setprecision(2) << coordinateTimeHours << " hrs\n"
                  << "   Proper Clock (τ): " << properTimeHours << " hrs (Time Deficit: +" << timeDeficit << " hrs)\n"
                  << "   Velocity        : " << currentVelocityKmS << " km/s (Lorentz γ: " << std::setprecision(4) << lorentzGamma << ")\n"
                  << "   Cumulative Δv   : " << std::setprecision(2) << cumulativeDeltaVKmS << " km/s\n"
                  << "   Distance Flown  : " << std::scientific << std::setprecision(3) << distanceKm << " km\n";
    }
};

std::cout << "[PROPAGATORS READY] HohmannTransferSolver and RelativisticKinematicsSolver active.\n";
````

## 7. Dynamic Polymorphic Mission Control Host
The `MissionControlHost` orchestrates all derived subsystems using polymorphic pointers (`std::shared_ptr<ISimulationModule>`).
It demonstrates runtime dynamic dispatch across heterogeneous modules through a single unified interface.


````c++
// ============================================================================
// ORCHESTRATOR CLASS: MissionControlHost
// Demonstrates Aggregation and Dynamic Polymorphic Dispatch
// ============================================================================
class MissionControlHost {
private:
    std::string missionCodename;
    std::vector<std::shared_ptr<ISimulationModule>> modules;

public:
    explicit MissionControlHost(std::string codename)
        : missionCodename(std::move(codename)) {}

    void registerModule(std::shared_ptr<ISimulationModule> mod) {
        if (mod) modules.push_back(mod);
    }

    void bootstrapAll() {
        std::cout << "=================================================================================\n";
        std::cout << "   MISSION CONTROL HOST: " << missionCodename << "\n";
        std::cout << "   BOOTSTRAPPING ALL REGISTERED SUBSYSTEMS (Dynamic Polymorphic Initialization)   \n";
        std::cout << "=================================================================================\n";
        for (auto& mod : modules) {
            std::cout << ">> Initializing: " << mod->getModuleIdentifier() << "...\n";
            mod->initialize(); // Dynamic dispatch (Method Overriding)
        }
        std::cout << ">> All " << modules.size() << " subsystems reporting NOMINAL status.\n\n";
    }

    void stepMission(double dtSec, int totalCycles) {
        for (int c = 1; c <= totalCycles; ++c) {
            for (auto& mod : modules) {
                mod->step(dtSec); // Dynamic dispatch
            }
        }
    }

    void dispatchAllDiagnostics() const {
        std::cout << "---------------------------------------------------------------------------------\n";
        std::cout << "                 REAL-TIME TELEMETRY STREAM & DIAGNOSTICS                        \n";
        std::cout << "---------------------------------------------------------------------------------\n";
        for (const auto& mod : modules) {
            mod->displayDiagnostics(); // Dynamic dispatch
            std::cout << "\n";
        }
        std::cout << "---------------------------------------------------------------------------------\n";
    }
};

// Instantiate Mission Control
auto mcc = std::make_unique<MissionControlHost>("MARS CITADEL EXPEDITION DECK 01");

// Create Concrete Derived Instances
auto lidar = std::make_shared<LidarSensor>(128, 65536, 145.0);
auto spec = std::make_shared<OpticalSpectrometer>(85.0);

// Use Method Overloading on Spectrometer (setting 4,200 km/s closing velocity)
spec->setRadialVelocity(4200.0);

auto dsn = std::make_shared<DSNTransceiver>(8.41e9, 35.0, 74.0, 110.0);
// Use Method Overloading on DSN (setting Mars distance in AU)
dsn->setTargetDistance(1.524);

auto hohmann = std::make_shared<HohmannTransferSolver>(1.000, 1.524, 1.32712440018e11);
auto relKinematics = std::make_shared<RelativisticKinematicsSolver>(5000.0, 1.32712440018e11);

// Polymorphic Registration
mcc->registerModule(lidar);
mcc->registerModule(spec);
mcc->registerModule(dsn);
mcc->registerModule(hohmann);
mcc->registerModule(relKinematics);

// Bootstrap, step simulation for 1 hour (360 cycles * 10 sec), and dispatch telemetry
mcc->bootstrapAll();
mcc->stepMission(10.0, 360);
mcc->dispatchAllDiagnostics();

std::cout << "Master System Verification Complete: All OOP Mechanisms Verified.\n";
std::cout << "Engineering Leadership: Ananya Gangwar, Yashvi Patel & Drashti\n";
std::cout << "Streamlit Live App: https://app-simulation-suite-utmetby2sfkmz3zcekutei.streamlit.app\n";
std::cout << "GitHub Repository : https://github.com/ananya39836-blip/streamlit-simulation-suite\n";
std::cout << "=================================================================================\n";
````
