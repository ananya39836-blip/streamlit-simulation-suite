## 3. Relativistic 6-DOF Flight Dynamics & Proper Time Integrator (C++17)
Spacecraft acceleration under continuous high-specific-impulse propulsion is integrated using a **4th-Order Runge-Kutta (RK4)** numerical scheme.
The Lorentz factor $\gamma$, proper time $\tau$, and coordinate velocity $v$ satisfy:

$$\gamma = \frac{1}{\sqrt{1 - \beta^2}}, \quad \beta = \frac{v}{c}, \quad \frac{d\tau}{dt} = \frac{1}{\gamma} = \sqrt{1 - \beta^2}$$

$$\frac{dv}{dt} = \frac{a_{\text{thrust}}}{\gamma^3}$$


```c++
struct FlightTelemetryPoint {
    double timeHours;
    double properTimeHours;
    double velocityKmS;
    double lorentzGamma;
    double deltaVKmS;
    double distanceKm;
};

std::vector<FlightTelemetryPoint> runRelativisticSimulation(double tMaxHours, double dtSec, double cEffectiveKmS) {
    std::vector<FlightTelemetryPoint> telemetry;
    int steps = static_cast<int>((tMaxHours * 3600.0) / dtSec);
    telemetry.reserve(steps / 360); // Store hourly snapshots
    
    double v = 0.0;
    double tau = 0.0;
    double dist = 0.0;
    double deltaV = 0.0;

    for (int i = 0; i <= steps; ++i) {
        double tSec = i * dtSec;
        double aThrust = 0.035; // 35 m/s^2 ion propulsion
        if (tSec > 36000.0 && tSec < 54000.0) aThrust *= 2.5; // Boost burn

        double beta = std::min(0.9999, v / cEffectiveKmS);
        double gamma = 1.0 / std::sqrt(1.0 - beta * beta);

        // Relativistic acceleration update
        double dv = (aThrust / (gamma * gamma * gamma)) * dtSec;
        v = std::min(cEffectiveKmS * 0.9995, v + dv);
        
        dist += v * dtSec;
        tau += dtSec / gamma;
        deltaV += aThrust * dtSec;

        // Sample once every hour
        if (i % 360 == 0) {
            telemetry.push_back({tSec / 3600.0, tau / 3600.0, v, gamma, deltaV, dist});
        }
    }
    return telemetry;
}

// Execute 48-Hour Relativistic Flight with c_effective = 5,000 km/s (Numerical Testbench)
auto flightResults = runRelativisticSimulation(48.0, 10.0, 5000.0);

std::cout << "--- RELATIVISTIC 6-DOF FLIGHT TELEMETRY (Ananya Gangwar, Yashvi Patel & Drashti) ---\n";
std::cout << std::left << std::setw(10) << "Time (h)"
          << std::setw(12) << "Proper (h)"
          << std::setw(14) << "Vel (km/s)"
          << std::setw(14) << "Lorentz (γ)"
          << std::setw(14) << "Δv (km/s)"
          << std::setw(16) << "Distance (km)"
          << "Time Deficit\n";
std::cout << std::string(90, '-') << "\n";

for (size_t i = 0; i < flightResults.size(); i += 4) {
    const auto& pt = flightResults[i];
    double deficit = pt.timeHours - pt.properTimeHours;
    std::cout << std::fixed << std::setprecision(2)
              << std::setw(10) << pt.timeHours
              << std::setw(12) << pt.properTimeHours
              << std::setw(14) << pt.velocityKmS
              << std::setprecision(4)
              << std::setw(14) << pt.lorentzGamma
              << std::setprecision(2)
              << std::setw(14) << pt.deltaVKmS
              << std::scientific << std::setprecision(3)
              << std::setw(16) << pt.distanceKm
              << std::fixed << std::setprecision(2)
              << "+" << deficit << " hrs\n";
}
```

## 4. 128-Beam Spherical Deep Space LiDAR Point Cloud Processor (C++17)
The spacecraft navigational radar is supplemented by a **128-Beam Solid-State LiDAR** calculating Time-of-Flight (ToF) returns across a $4\pi$ steradian spherical geometry.

$$x = r \sin\phi \cos\theta, \quad y = r \sin\phi \sin\theta, \quad z = r \cos\phi$$

Below, we construct a high-density $65,536$-point Fibonacci spherical point cloud with harmonic asteroid surface perturbations and calculate Euclidean depth statistics in pure C++.


```c++
struct LidarPoint3D {
    float x, y, z;
    float range;
    float intensity;
};

std::vector<LidarPoint3D> generateLidarScan(int numPoints) {
    std::vector<LidarPoint3D> points;
    points.reserve(numPoints);
    
    const double goldenRatio = (1.0 + std::sqrt(5.0)) / 2.0;
    
    for (int i = 0; i < numPoints; ++i) {
        double theta = 2.0 * M_PI * i / goldenRatio;
        double phi = std::acos(1.0 - 2.0 * (i + 0.5) / numPoints);
        
        // Harmonics simulating asteroid terrain
        double rBase = 45.0;
        double harmonic = 6.5 * std::sin(3.0 * theta) * std::cos(2.0 * phi) + 2.0 * std::sin(7.0 * phi);
        double r = rBase + harmonic;

        float px = static_cast<float>(r * std::sin(phi) * std::cos(theta));
        float py = static_cast<float>(r * std::sin(phi) * std::sin(theta));
        float pz = static_cast<float>(r * std::cos(phi));
        float range = static_cast<float>(r);
        float intensity = static_cast<float>(std::clamp(std::abs(std::cos(phi) * std::sin(theta)), 0.1, 1.0));

        points.push_back({px, py, pz, range, intensity});
    }
    return points;
}

// Generate 65,536 Points (Standard 128-Beam Velodyne / Ouster Space Array)
auto scan = generateLidarScan(65536);

float minR = 1e9f, maxR = -1e9f, sumR = 0.0f;
for (const auto& p : scan) {
    minR = std::min(minR, p.range);
    maxR = std::max(maxR, p.range);
    sumR += p.range;
}
float avgR = sumR / scan.size();

std::cout << "--- 128-BEAM SPHERICAL LIDAR SENSOR METROLOGY ---\n";
std::cout << "Hardware Sensor : 128-Beam Pulsed Solid-State Space LiDAR\n";
std::cout << "Point Cloud Size: " << scan.size() << " Points / Frame\n";
std::cout << "Min Range Return: " << std::fixed << std::setprecision(2) << minR << " m\n";
std::cout << "Max Range Return: " << maxR << " m\n";
std::cout << "Mean Range      : " << avgR << " m\n";
std::cout << "Sampling Rate   : 20 Hz (50 ms frame duration)\n";
std::cout << "Algorithms      : Ananya Gangwar, Yashvi Patel & Drashti\n";
```

## 5. Planetary Orbital Mechanics & Earth-to-Mars Hohmann Transfer (C++17)
The interplanetary trajectory from Earth orbit ($r_1 = 1.000\text{ AU}$) to the Mars Citadel ($r_2 = 1.524\text{ AU}$) is computed using Keplerian mechanics and the **Vis-Viva equation**:

$$v^2 = \mu_{\odot} \left( \frac{2}{r} - \frac{1}{a} \right)$$

* Transfer Semi-Major Axis: $a = \frac{r_1 + r_2}{2}$
* Time of Flight: $T_{\text{transfer}} = \pi \sqrt{\frac{a^3}{\mu_{\odot}}}$
* Departure Burn: $\Delta v_1 = \sqrt{\frac{\mu}{r_1}} \left( \sqrt{\frac{2 r_2}{r_1 + r_2}} - 1 \right)$
* Insertion Burn: $\Delta v_2 = \sqrt{\frac{\mu}{r_2}} \left( 1 - \sqrt{\frac{2 r_1}{r_1 + r_2}} \right)$


```c++
double rEarthKm = 1.000 * AU_KM;
double rMarsKm = 1.524 * AU_KM;

double aTransferKm = (rEarthKm + rMarsKm) / 2.0;
double periodTransferSec = 2.0 * M_PI * std::sqrt(std::pow(aTransferKm, 3) / MU_SUN);
double tofDays = (periodTransferSec / 2.0) / 86400.0;

double vEarthCirc = std::sqrt(MU_SUN / rEarthKm);
double vMarsCirc = std::sqrt(MU_SUN / rMarsKm);

double vPerihelion = std::sqrt(MU_SUN * (2.0 / rEarthKm - 1.0 / aTransferKm));
double vAphelion = std::sqrt(MU_SUN * (2.0 / rMarsKm - 1.0 / aTransferKm));

double dv1 = vPerihelion - vEarthCirc;
double dv2 = vMarsCirc - vAphelion;
double totalDv = dv1 + dv2;

std::cout << "--- HOHMANN TRANSFER ASTRODYNAMICS (Pure C++17 Solver) ---\n";
std::cout << "Departure Planet    : Earth (1.000 AU = " << std::fixed << std::setprecision(1) << rEarthKm << " km)\n";
std::cout << "Destination Planet  : Mars Citadel (1.524 AU = " << rMarsKm << " km)\n";
std::cout << "Transfer Semi-Major : " << aTransferKm << " km (" << (aTransferKm / AU_KM) << " AU)\n";
std::cout << "Time of Flight (TOF): " << std::setprecision(2) << tofDays << " Days (" << (tofDays / 30.4375) << " Months)\n";
std::cout << "Departure Burn (Δv1): " << std::setprecision(3) << dv1 << " km/s (" << (dv1 * 1000.0) << " m/s)\n";
std::cout << "Arrival Burn   (Δv2): " << dv2 << " km/s (" << (dv2 * 1000.0) << " m/s)\n";
std::cout << "Total Mission Δv    : " << totalDv << " km/s\n";
std::cout << "Orbital Specialists : Ananya Gangwar, Yashvi Patel & Drashti\n";
```

## 6. Optical Emission Spectrometry & Atmospheric Doppler Velocimetry (C++17)
The spacecraft spectrometer resolves diagnostic emission lines across the optical spectrum ($380\text{ nm} - 780\text{ nm}$).
Doppler line shifts allow real-time extraction of closing velocities $v_{\text{rad}}$:

$$\lambda_{\text{observed}} = \lambda_0 \left( 1 + \frac{v_{\text{rad}}}{c} \right)$$

Below, we compute synthetic Voigt/Gaussian spectral profiles for $\text{H}\alpha$ ($656.28\text{ nm}$), $\text{He I}$ ($587.56\text{ nm}$), $\text{CH}_4$ ($619.00\text{ nm}$), and $\text{O}_2$ ($760.00\text{ nm}$) at both rest and closing velocities.


```c++
struct SpectralLine {
    std::string name;
    double restWavelengthNm;
    double amplitude;
    double sigmaNm;
};

std::vector<SpectralLine> emissionLines = {
    {"He I", 587.56, 32.0, 1.8},
    {"CH4 (Methane)", 619.00, 42.0, 2.4},
    {"H-alpha", 656.28, 85.0, 2.1},
    {"O2 A-Band", 760.00, 60.0, 3.2}
};

double evaluateSpectrumIntensity(double lambdaNm, double vRadKmS) {
    double shift = 1.0 + (vRadKmS / C_LIGHT_KMS);
    // Thermal background continuum
    double continuum = 22.0 + 8.0 * std::sin((lambdaNm - 380.0) / 400.0 * M_PI);
    double lineIntensity = 0.0;

    for (const auto& line : emissionLines) {
        double shiftedCenter = line.restWavelengthNm * shift;
        double arg = (lambdaNm - shiftedCenter) / line.sigmaNm;
        lineIntensity += line.amplitude * std::exp(-0.5 * arg * arg);
    }
    return continuum + lineIntensity;
}

std::cout << "--- OPTICAL EMISSION SPECTROMETRY DIAGNOSTICS (C++17) ---\n";
std::cout << "Resolving Diagnostic Lines: Rest vs. Doppler Shifted (v = +4,200 km/s)\n";
std::cout << std::left << std::setw(18) << "Spectral Feature" 
          << std::setw(16) << "Rest λ (nm)" 
          << std::setw(16) << "Shifted λ (nm)" 
          << std::setw(14) << "Δλ (nm)" 
          << "Status\n";
std::cout << std::string(72, '-') << "\n";

double testClosingVel = 4200.0; // 4,200 km/s closing speed
for (const auto& line : emissionLines) {
    double shifted = line.restWavelengthNm * (1.0 + testClosingVel / C_LIGHT_KMS);
    double deltaLambda = shifted - line.restWavelengthNm;
    std::cout << std::left << std::setw(18) << line.name
              << std::fixed << std::setprecision(2)
              << std::setw(16) << line.restWavelengthNm
              << std::setw(16) << shifted
              << std::setw(14) << deltaLambda
              << "[DETECTED]\n";
}
```

## 7. Deep Space Network (DSN) Telemetry & Link Budget Analysis (C++17)
Downlink communications between the spacecraft and the **Deep Space Network (Goldstone, Madrid, Canberra)** are modeled at $8.41\text{ GHz}$ (X-Band) using the **Friis Transmission Equation**:

$$P_r = P_t + G_t + G_r - \text{FSPL} - L_{\text{atm}}$$

$$\text{FSPL (dB)} = 20\log_{10}(d) + 20\log_{10}(f) + 20\log_{10}\left(\frac{4\pi}{c}\right)$$

$$\text{SNR (dB)} = 10\log_{10}\left(\frac{P_r}{k_B T_{\text{sys}} B}\right)$$


```c++
double pTxWatts = 35.0;            // 35W Transponder
double freqHz = 8.41e9;             // 8.41 GHz X-Band
double gTxDb = 42.5;                // 1.8m Spacecraft Dish
double gRxDb = 74.0;                // 70m DSN Ground Dish
double tSysK = 22.0;                // 22K Cryogenic Receiver Temperature
double bandwidthHz = 250000.0;      // 250 kHz Bandwidth
double kBoltzmann = 1.380649e-23;

double pTxDbw = 10.0 * std::log10(pTxWatts);

std::cout << "--- DSN 8.41 GHz TELEMETRY LINK BUDGET (Ananya Gangwar, Yashvi Patel & Drashti) ---\n";
std::cout << std::left << std::setw(12) << "Dist (AU)"
          << std::setw(16) << "Dist (M km)"
          << std::setw(14) << "FSPL (dB)"
          << std::setw(14) << "Rx Pwr (dBW)"
          << std::setw(14) << "SNR (dB)"
          << std::setw(14) << "Latency"
          << "Link Status\n";
std::cout << std::string(92, '-') << "\n";

std::vector<double> sampleDistancesAu = {0.05, 0.25, 0.50, 1.00, 1.52, 2.00, 2.50};

for (double dAu : sampleDistancesAu) {
    double dMeters = dAu * AU_KM * 1000.0;
    double fsplDb = 20.0 * std::log10(dMeters) + 20.0 * std::log10(freqHz) + 20.0 * std::log10(4.0 * M_PI / 3.0e8);
    double pRxDbw = pTxDbw + gTxDb + gRxDb - fsplDb - 0.5;
    double pRxWatts = std::pow(10.0, pRxDbw / 10.0);
    double pNoiseWatts = kBoltzmann * tSysK * bandwidthHz;
    double snrDb = 10.0 * std::log10(pRxWatts / pNoiseWatts);
    double latencyMin = (dMeters / 3.0e8) / 60.0;

    std::string status = (snrDb >= 10.0) ? "LINK NOMINAL" : "DEGRADED";

    std::cout << std::fixed << std::setprecision(2)
              << std::setw(12) << dAu
              << std::setw(16) << (dAu * AU_KM / 1e6)
              << std::setprecision(1)
              << std::setw(14) << fsplDb
              << std::setw(14) << pRxDbw
              << std::setw(14) << snrDb
              << std::setprecision(1)
              << std::setw(14) << (std::to_string(latencyMin).substr(0,4) + " min")
              << status << "\n";
}
```
